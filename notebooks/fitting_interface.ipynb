```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output
from pathlib import Path
from src.processing import limpiar_datos, filtrar_datos
from src.plotting import graficar_analisis

# Cargar datos
df = pd.read_parquet('results/dataset_clasificado.parquet')

# Widgets
fz_dropdown = widgets.Dropdown(options=sorted(df['FZ_nom'].unique()), description='FZ_nom:')
p_dropdown = widgets.Dropdown(options=sorted(df['P_nom'].unique()), description='P_nom:')
ia_dropdown = widgets.Dropdown(options=sorted(df['IA_nom'].unique()), description='IA_nom:')
plot_button = widgets.Button(description="Actualizar Plot")
output = widgets.Output()

def update_plot(_):
    with output:
        clear_output()
        criterios = {'FZ_nom': fz_dropdown.value, 'P_nom': p_dropdown.value, 'IA_nom': ia_dropdown.value}
        df_filt = filtrar_datos(df, criterios)
        df_clean = limpiar_datos(df_filt)
        
        if not df_clean.empty:
            print(f"Ploteando {len(df_clean)} puntos.")
            graficar_analisis(df_clean, Path('plots/temp'), 'SA', 'FY', 'FZ_nom')
        else:
            print("No hay datos con esos criterios.")

plot_button.on_click(update_plot)
display(fz_dropdown, p_dropdown, ia_dropdown, plot_button, output)
```
